```python
from sklearn.metrics import cohen_kappa_score
import pandas as pd
```

```python
df_new = pd.read_csv("./data/GoldSet100.csv")
df_old = pd.read_csv("./data/GoldSet.csv")
```

```python
df = pd.merge(df_new, df_old, on=['Diff', "File", "Project", "CommitHash", "CommitMessage"], how='inner', suffixes=('_s', '_n'))

label_s = df["Decision_s"].tolist()
label_n = df["Decision_n"].tolist()
```

```python
cohen_kappa_score(label_s, label_n)
```
